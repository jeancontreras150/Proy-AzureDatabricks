```python

spark.sql(f"""CREATE OR REPLACE TABLE {PRM_catalogo}.bronze.Clientes_Sistema	
(
User_id STRING,
Last_Name STRING,
Name STRING,
Age INT,
Cell_number STRING,
Sign_date DATE,
Fecha_proceso TIMESTAMP
)
USING DELTA
LOCATION 'abfss://cont-ecom@{PRM_nameStorage}.dfs.core.windows.net/bronze/tablas/Clientes_Sistema';
""")

spark.sql(f"""CREATE OR REPLACE TABLE {PRM_catalogo}.bronze.Interaccion_Sistema		
(
TypeInt_id STRING,	
Interaction_name STRING,	
DestipoInt_Spanish STRING,
Registration_system_date STRING,
Fecha_proceso TIMESTAMP
)
USING DELTA
LOCATION 'abfss://cont-ecom@{PRM_nameStorage}.dfs.core.windows.net/bronze/tablas/Interaccion_Sistema';
""")

spark.sql(f"""CREATE OR REPLACE TABLE {PRM_catalogo}.bronze.Productos_Sistema		
(
Product_id STRING,
Producto_Name STRING,
Category STRING,
Price DOUBLE,
Fecha_proceso TIMESTAMP
)
USING DELTA
LOCATION 'abfss://cont-ecom@{PRM_nameStorage}.dfs.core.windows.net/bronze/tablas/Productos_Sistema';
""")

spark.sql(f"""CREATE OR REPLACE TABLE {PRM_catalogo}.bronze.ecommerce_data		
(
Interaction_id_system VARCHAR(6),
Interaction_date STRING,
User_id_system STRING,
Product_id STRING,
TypeInt_id VARCHAR(30),
Event VARCHAR(256),
Location STRING,
Quantity STRING,
Product_rating DECIMAL(10,1),
Fecha_proceso TIMESTAMP
)
USING DELTA
LOCATION 'abfss://cont-ecom@{PRM_nameStorage}.dfs.core.windows.net/bronze/tablas/ecommerce_data';
""")
```
