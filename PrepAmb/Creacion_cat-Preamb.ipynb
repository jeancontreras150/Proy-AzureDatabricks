```python
## PARAMETRIZAR ADLS Y CATALOGO A PROD
dbutils.widgets.text("catalogo","catalogo_desa_intEcommerce")
dbutils.widgets.text("storageName","adlsecompag")
dbutils.widgets.text("ambiente","desa")

catalogo = dbutils.widgets.get("catalogo")
storageName = dbutils.widgets.get("storageName")
ambiente = dbutils.widgets.get("ambiente")
```

```python
spark.sql(f"DROP CATALOG IF EXISTS {catalogo} CASCADE;")
```

```python

spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalogo};")
```

```python

##CREACION DE ESQUEMAS PARA LAS TABLAS FISICAS
spark.sql(f"""CREATE SCHEMA IF NOT EXISTS {catalogo}.bronze;""")
spark.sql(f"""CREATE SCHEMA IF NOT EXISTS {catalogo}.silver;""")
spark.sql(f"""CREATE SCHEMA IF NOT EXISTS {catalogo}.golden;""")
##CREACION DE ESQUEMAS VISTAS PARA USUARIOS SIN ACCESO A DATOS CRITICOS
spark.sql(f"""CREATE SCHEMA IF NOT EXISTS {catalogo}.bronze_v;""")
spark.sql(f"""CREATE SCHEMA IF NOT EXISTS {catalogo}.silver_v;""")
spark.sql(f"""CREATE SCHEMA IF NOT EXISTS {catalogo}.golden_v;""")
##CREACION DE ESQUEMA VISTAS PARA USUARIOS CON ACCESO A DATOS CRITICOS
spark.sql(f"""CREATE SCHEMA IF NOT EXISTS {catalogo}.bronze_vdc;""")
spark.sql(f"""CREATE SCHEMA IF NOT EXISTS {catalogo}.silver_vdc;""")
spark.sql(f"""CREATE SCHEMA IF NOT EXISTS {catalogo}.golden_vdc;""")
```

```python
#CREACION EXTERNAL LOCATION DE LOS INSUMOS
spark.sql(f"""CREATE EXTERNAL LOCATION IF NOT EXISTS `exlt-raw-insumos-{ambiente}`
URL 'abfss://raw-insumos@{storageName}.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL credential)
COMMENT 'Ubicación externa del container de los insumos del Data Lake';
""")
```

```python

##CREACION EXTERNAL LOCATION DE LAS TABLAS DE LAS CAPAS MEDALLION
spark.sql(f"""CREATE EXTERNAL LOCATION IF NOT EXISTS `exlt-cont-ecom-{ambiente}`
URL 'abfss://cont-ecom@{storageName}.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL credential)
COMMENT 'Ubicación externa para las tablas unit-catalog del Data Lake del proyecto';
""")
```
